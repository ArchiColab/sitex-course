# Interactive Webmap — Gehl Activities & Third Space

**Library:** [`sitex`](../sitex) · **Original, fully-documented version:** [`04-VIZ-InteractiveWebmap-Gehl.ipynb`](../documentations/04-VIZ-InteractiveWebmap-Gehl.ipynb)

The plain categorical point-data counterpart to the KDE density view
(`04-VIZ-POIsHeatmaps.ipynb`): every classified POI, coloured by Gehl activity /
Third Space, over grey building footprints on a dark basemap — individual places stay
identifiable and clickable rather than blurred into a density surface. Already pure
folium in the original notebook — no matplotlib rendering to work around here.

```python
# One-time setup, if `sitex` isn't already installed in this kernel:
# %pip install "sitex[viz] @ git+https://github.com/ArchiColab/sitex.git"

import sys
from pathlib import Path

IN_COLAB = "google.colab" in sys.modules
print("Running in:", "Google Colab" if IN_COLAB else "Local environment")

if IN_COLAB:
    %pip install "sitex[viz] @ git+https://github.com/ArchiColab/sitex.git"

    # Input data (read-only) — downloaded fresh each session from the workshop's GitHub release
    DATA_RELEASE_URL = "https://github.com/ArchiColab/sitex/releases/download/workshop-data-v1/Colab_Outputs.zip"
    DATA_DIR = Path("/content/Colab_Outputs")
    if not DATA_DIR.exists():
        import urllib.request, zipfile
        zip_path = Path("/content/Colab_Outputs.zip")
        urllib.request.urlretrieve(DATA_RELEASE_URL, zip_path)
        with zipfile.ZipFile(zip_path) as zf:
            zf.extractall(DATA_DIR)

    # Your own results — saved to your own Google Drive so they persist across sessions
    from google.colab import drive
    drive.mount("/content/gdrive")
    OUTPUT_DIR = Path("/content/gdrive/MyDrive/SiteX_Outputs")
else:
    DATA_DIR = Path("..") / "data"
    OUTPUT_DIR = Path("..") / "outputs"

DATA_DIR.mkdir(parents=True, exist_ok=True)
OUTPUT_DIR.mkdir(parents=True, exist_ok=True)

import geopandas as gpd

from sitex.core.config import CityConfig
from sitex.viz import interactive_webmap as iw
```

```python
city = CityConfig(
    place_name="Phường Pleiku, Gia Lai, Vietnam",
    local_lat=13.9833,
    local_lon=108.0000,
    slug="phường_pleiku",
    data_dir=DATA_DIR,
    output_dir=OUTPUT_DIR,
)
CITY_KEY = "pleiku"  # the classified-POIs backup's own city label — see 03-NA02/03's notes

CATEGORY_STYLE = {
    "Necessary": {"col": "is_necessary", "color": "#3182bd"},
    "Optional": {"col": "is_optional", "color": "#31a354"},
    "Social": {"col": "is_social", "color": "#e6550d"},
    "Third Space": {"col": "is_thirdspace", "color": "#756bb1"},
}

pois = gpd.read_file(city.data_dir / "overture" / f"pois_classified_{CITY_KEY}.gpkg")
pois_web = pois.to_crs("EPSG:4326")
for name, spec in CATEGORY_STYLE.items():
    print(f"{name:<12s} -> {int(pois_web[spec['col']].sum()):,} POIs")

SIMPLIFY_TOLERANCE_M = 0.5
buildings = gpd.read_file(city.data_dir / "overture" / f"{city.slug}_buildings.gpkg").to_crs(pois.crs)
buildings["geometry"] = buildings.geometry.simplify(SIMPLIFY_TOLERANCE_M)
buildings_web = buildings.to_crs("EPSG:4326")
print(f"{len(buildings_web):,} building footprints")
```

```python
webmap = iw.build_gehl_point_webmap((city.local_lat, city.local_lon), pois_web, buildings_web, CATEGORY_STYLE)

webmap_path = city.output_dir / f"{CITY_KEY}_webmap_gehl_thirdspace.html"
webmap.save(str(webmap_path))
print(f"Webmap saved to: {webmap_path}")
print(f"File size: {webmap_path.stat().st_size / 1e6:.1f} MB")
webmap
```

## Notes

- All layers are on by default here — POI counts are small enough that there's no
  load-time cost to showing everything at once. Toggle any off from the layer control.
- A POI can qualify for more than one category (e.g. a cafe is often both Optional and
  Social) — it gets a marker in every layer it qualifies for.
