# Program — POIs to Functional Group, Gehl Activity & Public Amenity

**Library:** [`sitex`](../sitex) · **Original, fully-documented version:** [`01-ARCH-POIsClassification.ipynb`](../documentations/01-ARCH-POIsClassification.ipynb)

Thin, parameterized version of the POI classification workflow. This is pure table
lookups — **no live LLM.** The two inputs it joins come from different places on purpose:

| Input | What it is |
|---|---|
| `data/overture/{slug}_places.gpkg` | This AOI's own Overture places, extracted by `00-Data-Acquisition.ipynb` (Section 8) — any OSM-geocodable place, not a fixed city list |
| `data/reference/overture_place_classification.json` | Every Overture category → functional group + Gehl activity, frozen into a lookup table by the background `00- Overture - Activities Classifier.ipynb` notebook |

The classification lookup is **place-independent** — it's keyed purely by Overture category
slug (`restaurant` → Food & Drink, regardless of which city that restaurant is in). It was built
once from a handful of big Vietnamese cities just to get broad category coverage, not to restrict
which places this notebook can classify. Any category it hasn't seen before defaults to
`Unclassified` rather than erroring, so a new AOI is always safe to run.

All the logic lives in `sitex.arch.poi_classification` — this notebook is the
parameters plus the calls that use them.

## 2. Site configuration

`slug` is set explicitly here to match the filenames `00-Data-Acquisition.ipynb` already
wrote to `data/overture/` for this AOI (its own place-string slug, diacritics included —
same convention `01-ARCH-Building_Morphology.ipynb` §2 uses for buildings/streets). Use
the same `place_name`/`local_lat`/`local_lon`/`slug` you used there, for any OSM-geocodable
place.

```python
city = CityConfig(
    place_name="Phường Pleiku, Gia Lai, Vietnam",
    local_lat=13.9833,
    local_lon=108.0000,
    slug="phường_pleiku",  # matches 00-Data-Acquisition.ipynb's own filename slug
    data_dir=DATA_DIR,
    output_dir=OUTPUT_DIR,
)

print(f"AOI slug: {city.slug}")
print(f"Local EPSG (for the reprojected export in Section 6): {city.local_epsg}")
```

## 3. Load the classification lookup

`is_public_amenity` answers a different question from Gehl activity: not *why* someone
visits (Necessary/Optional/Social — behavior), but whether an accessibility study
(isochrones to healthcare, education, ...) should treat this category as an essential
public/civic service. A hospital is both Necessary *and* a public amenity; a private gym
is Optional and not one.

```python
lookup = poi.load_classification_lookup_for_city(city)
category_to_l1 = poi.load_taxonomy_l1_lookup_for_city(city)
lookup = poi.add_public_amenity_flag(lookup, category_to_l1)

print(f"{len(lookup):,} categories in the classification lookup")
print(f"{lookup['is_public_amenity'].sum():,} flagged as public amenities:")
print(lookup.loc[lookup['is_public_amenity'], 'functional_group'].value_counts().to_string())
```

## 4. Load places & classify

```python
places = poi.load_places_for_aoi(city)
print(f"{len(places):,} places loaded for '{city.slug}'")

pois, missing = poi.classify_places(places, lookup)
if len(missing):
    print(f"{len(missing)} categories not in the reference table (defaulted to Unclassified):")
    print(missing.to_string())
else:
    print("Every category in this AOI is covered by the reference table.")
```

## 5. Gehl activity flags

A POI can support more than one Gehl activity (e.g. a cafe is both Optional and Social) —
each gets its own boolean flag rather than being forced into one exclusive category.

```python
pois = poi.add_gehl_activity_flags(pois)
pois_final = poi.select_final_columns(pois)

print(f"{len(pois_final):,} classified POIs\n")
print("—— Functional group ——")
print(pois_final["functional_group"].value_counts().to_string())
print(f"\nNecessary: {pois_final['is_necessary'].sum():,}  "
      f"Optional: {pois_final['is_optional'].sum():,}  "
      f"Social: {pois_final['is_social'].sum():,}")
print(f"Public amenities: {pois_final['is_public_amenity'].sum():,} / {len(pois_final):,}")
print(f"Candidate third places (first-pass, review before trusting): "
      f"{pois_final['is_thirdspace'].sum():,} / {len(pois_final):,}")
```

## 6. Map & export

The GeoJSON export stays in EPSG:4326 (GeoJSON convention, what the map needs). The GPKG
is reprojected to `city.local_epsg` — required for any spatial join against buildings
(`01-ARCH-Building_Morphology.ipynb`) or Space Syntax outputs, which are already metric;
joining the WGS84 GeoJSON instead silently distorts distances.

```python
poi.plot_pois_gehl_map(pois_final)
```

```python
geojson_path = poi.export_classified_pois_geojson_for_city(pois_final, city)
print(f"GeoJSON (WGS84)        → {geojson_path}")

gpkg_path = poi.export_classified_pois_gpkg_for_city(pois_final, city)
print(f"GPKG (EPSG:{city.local_epsg})    → {gpkg_path}")
print("Ready for the next phase: isochrone / service-area / accessibility analysis.")
```

---
## Notes

- **`is_thirdspace` is a first-pass, unvalidated rule** — review before trusting, per the
  original notebook's own caveat.
- **Use the `.gpkg`, not the `.geojson`, for any spatial join** against buildings or
  street-network outputs — see Section 6 above.
- **Run `00-Data-Acquisition.ipynb` first** for this AOI — this notebook fails fast
  (`FileNotFoundError`) if `{slug}_places.gpkg` isn't already on disk under
  `data/overture/`.
- To classify a new AOI: change `place_name`, `local_lat`/`local_lon`, and `slug` in
  Section 2 to match whatever you used in `00-Data-Acquisition.ipynb` — any
  OSM-geocodable place works, not only the cities sampled to build the classification
  lookup. `city.local_epsg` is re-derived automatically (Vietnam spans UTM zones 48N and
  49N, split at 108°E).
